```python
X = df.drop('Class' , axis=1)
```

```python
y = df['Class']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=101)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
help(RandomForestClassifier)
```

```python
n_estimators = [64,100,128,200]
```

```python
max_features = [2,3,4]
```

```python
bootstrap = [True,False]
```

```python
oob_score = [True,False]
```

```python
param_grid = {'n_estimators':n_estimators , 'max_features':max_features , 'bootstrap':bootstrap , 'oob_score':oob_score}
```

```python
rfc = RandomForestClassifier()
```

```python
grid = GridSearchCV(rfc , param_grid)
```

```python
grid.fit(X_train , y_train)
```

```python
grid.best_params_
```

```python
rfc = RandomForestClassifier(max_features=2 , n_estimators=64 , oob_score=True , bootstrap=True)
```

```python
rfc.fit(X_train , y_train)
```

```python
rfc.oob_score_
```

```python
predictions = grid.predict(X_test)
```

```python
from sklearn.metrics import classification_report , ConfusionMatrixDisplay , accuracy_score , confusion_matrix 
```

```python
confusion_matrix(y_test , predictions)
```

```python
print(classification_report(y_test , predictions))
```

```python
ConfusionMatrixDisplay.from_predictions(y_test , predictions)
```

```python
errors = []
misclassifications = []

for n in range(1,200):
    rfc = RandomForestClassifier(n_estimators=n,max_features=2)
    rfc.fit(X_train,y_train)
    preds = rfc.predict(X_test)
    err = 1-accuracy_score(y_test , preds)
    n_missed = np.sum(preds != y_test)
    errors.append(err)
    misclassifications.append(n_missed)
```

```python
plt.plot(range(1,200) , errors)
```

```python
plt.plot(range(1,200) , misclassifications)
```
